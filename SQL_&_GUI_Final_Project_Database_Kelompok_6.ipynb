```python
conn = sqlite3.connect('perbankan.db')
c = conn.cursor()

# Mengaktifkan dukungan foreign key
c.execute('PRAGMA foreign_keys = ON')

c.execute('''CREATE TABLE Nasabah (
             ID_Nasabah CHAR(4) PRIMARY KEY NOT NULL,
             Nama_Nasabah TEXT,
             NoTelp_Nasabah CHAR,
             Alamat_Nasabah CHAR,
             TTL_Nasabah CHAR(10))''')

c.execute('''CREATE TABLE Rekening_Bank (
             No_Rekening INTEGER PRIMARY KEY NOT NULL,
             Pin INTEGER NOT NULL,
             ID_Nasabah CHAR(4) NOT NULL,
             ID_Akun CHAR(4) NOT NULL)''')

c.execute('''CREATE TABLE Akun_MBanking (
             ID_Akun CHAR(4) PRIMARY KEY NOT NULL,
             Username CHAR(10),
             Password CHAR(6))''')

c.execute('''CREATE TABLE Transaksi (
             ID_Transaksi CHAR(4) PRIMARY KEY NOT NULL,
             ID_Akun CHAR(4) NOT NULL,
             ID_Jenis_Transaksi CHAR(2) NOT NULL,
             Tanggal_Transaksi CHAR(10) NOT NULL,
             Nominal_Transaksi INTEGER,
             ID_Biaya_Admin CHAR(2))''')

c.execute('''CREATE TABLE Biaya_Admin (
             ID_Biaya_Admin CHAR(2) PRIMARY KEY NOT NULL,
             Jenis_Biaya_Admin CHAR NOT NULL,
             Nominal_Biaya_Admin INTEGER)''')

c.execute('''CREATE TABLE Jenis_Transaksi (
             ID_Jenis_Transaksi CHAR(2) PRIMARY KEY NOT NULL,
             Jenis_Transaksi CHAR NOT NULL)''')

conn.commit()
conn.close()
```

## Implementasi Join & Select

```python
#Data dengan Nominal Transaksi diatas 150.000

import sqlite3
from tabulate import tabulate

# Hubungkan ke database SQLite
conn = sqlite3.connect('perbankan.db')
c = conn.cursor()

# Menjalankan query untuk mengambil/SELECT data dengan JOIN antara tabel
query = """
SELECT Nasabah.Nama_Nasabah, Rekening_Bank.ID_Akun, Nasabah.ID_Nasabah, Transaksi.ID_Transaksi, Transaksi.Nominal_Transaksi
FROM Transaksi
JOIN Rekening_Bank ON Transaksi.ID_Akun = Rekening_Bank.ID_Akun
JOIN Nasabah ON Rekening_Bank.ID_Nasabah = Nasabah.ID_Nasabah
WHERE Transaksi.Nominal_Transaksi > 150000;
"""
c.execute(query)

# Mengambil semua baris hasil query
rows = c.fetchall()

# Menampilkan hasil query dalam bentuk tabel
headers = ["Nama_Nasabah", "ID_Akun", "ID_Nasabah", "ID_Transaksi", "Nominal_Transaksi"]
print("Data dengan Nominal Transaksi diatas 150.000:")
print(tabulate(rows, headers, tablefmt="github"))

# Menutup koneksi
conn.close()
```

```python
#Data maksimal Transaksi Tiap Nasabah

import sqlite3
from tabulate import tabulate

# Hubungkan ke database SQLite
conn = sqlite3.connect('perbankan.db')
c = conn.cursor()

# Menjalankan query untuk mengambil data dengan max(Nominal_Transaksi) dari setiap Nama_Nasabah
query = """
WITH MaxTransaksi AS (
    SELECT Rekening_Bank.ID_Nasabah, MAX(Transaksi.Nominal_Transaksi) AS Max_Nominal_Transaksi
    FROM Transaksi
    JOIN Rekening_Bank ON Transaksi.ID_Akun = Rekening_Bank.ID_Akun
    GROUP BY Rekening_Bank.ID_Nasabah
)
SELECT Nasabah.Nama_Nasabah, Rekening_Bank.ID_Akun, Nasabah.ID_Nasabah, Transaksi.ID_Transaksi, Transaksi.Nominal_Transaksi
FROM MaxTransaksi
JOIN Rekening_Bank ON MaxTransaksi.ID_Nasabah = Rekening_Bank.ID_Nasabah
JOIN Transaksi ON Rekening_Bank.ID_Akun = Transaksi.ID_Akun AND
MaxTransaksi.Max_Nominal_Transaksi = Transaksi.Nominal_Transaksi
JOIN Nasabah ON Rekening_Bank.ID_Nasabah = Nasabah.ID_Nasabah;
"""
c.execute(query)

# Mengambil semua baris hasil query
rows = c.fetchall()

# Menampilkan hasil query dalam bentuk tabel
headers = ["Nama_Nasabah", "ID_Akun", "ID_Nasabah", "ID_Transaksi", "Nominal_Transaksi"]
print("Data maksimal Transaksi Tiap Nasabah:")
print(tabulate(rows, headers, tablefmt="github"))

# Menutup koneksi
conn.close()
```

## Masukkan Data Transaksi

```python
# Membuat widget input
id_transaksi_input = widgets.Text(
    value='',
    description='ID Trans:',
    disabled=False)

id_akun_input = widgets.Text(
    value='',
    description='ID Akun:',
    disabled=False)

id_jenis_transaksi_input = widgets.Dropdown(
    options=['T0', 'T1', 'T2', 'T3', 'T4'],
    value='T0',
    description='ID Jenis Trans:',
    disabled=False)

tanggal_transaksi_input = widgets.DatePicker(
    description='Tgl Trans:',
    disabled=False,
    value=datetime.now())

nominal_transaksi_input = widgets.IntText(
    value=0,
    description='Nominal:',
    disabled=False)

id_biaya_admin_input = widgets.Text(
    value='',
    description='Biaya Admin:',
    disabled=False)

submit_button = widgets.Button(
    description='Masukkan Data',
    disabled=False,
    button_style='success',
    tooltip='Klik untuk memasukkan data transaksi')

def insert_data(b):
    with sqlite3.connect('perbankan.db') as conn:
        c = conn.cursor()
        # Mengaktifkan foreign key
        c.execute('PRAGMA foreign_keys = ON')
        try:
            c.execute('''INSERT INTO Transaksi (ID_Transaksi, ID_Akun, ID_Jenis_Transaksi,
                         Tanggal_Transaksi, Nominal_Transaksi, ID_Biaya_Admin)
                         VALUES (?, ?, ?, ?, ?, ?)''',
                      (id_transaksi_input.value, id_akun_input.value,
                       id_jenis_transaksi_input.value, tanggal_transaksi_input.value.
                       strftime('%d-%m-%Y'), nominal_transaksi_input.value,
                       id_biaya_admin_input.value))
            print("Data transaksi berhasil dimasukkan.")
        except sqlite3.IntegrityError as e:
            print("Gagal memasukkan data: ", e)
        except Exception as e:
            print("Terjadi kesalahan: ", e)

submit_button.on_click(insert_data)

display(id_transaksi_input, id_akun_input, id_jenis_transaksi_input, tanggal_transaksi_input,
        nominal_transaksi_input, id_biaya_admin_input, submit_button)
```

## Cek Mutasi Berdasarkan Tanggal

```python
# Widget untuk input username dan password
username_input = widgets.Text(
    value='',
    description='Username:',
    style={'description_width': 'initial'})

password_input = widgets.Password(
    value='',
    description='Password:',
    style={'description_width': 'initial'})

login_button = widgets.Button(
    description='Login',
    button_style='success',
    tooltip='Klik untuk login')

output_login = widgets.Output()

# Widget untuk input tanggal awal dan akhir
tanggal_awal_widget = widgets.DatePicker(
    description='Tanggal Awal:',
    style={'description_width': 'initial'},
    layout={'visibility': 'hidden'})  # Awalnya tidak terlihat

tanggal_akhir_widget = widgets.DatePicker(
    description='Tanggal Akhir:',
    style={'description_width': 'initial'},
    layout={'visibility': 'hidden'})  # Awalnya tidak terlihat

cek_mutasi_button = widgets.Button(
    description='Cek Mutasi',
    button_style='info',
    tooltip='Cek mutasi berdasarkan rentang tanggal',
    layout={'visibility': 'hidden'})  # Awalnya tidak terlihat

output_mutasi = widgets.Output()

def validasi_user(username, password):
    conn = sqlite3.connect('perbankan.db')
    c = conn.cursor()
    c.execute('SELECT COUNT(*) FROM Akun_MBanking WHERE Username = ? AND Password = ?', (username, password))
    hasil = c.fetchone()[0]
    conn.close()
    return hasil > 0

# Fungsi untuk menangani login
def login(b):
    with output_login:
        clear_output()
        username = username_input.value
        password = password_input.value

        if validasi_user(username, password):
            print("Login berhasil. Silakan pilih rentang tanggal untuk cek mutasi.")
            tanggal_awal_widget.layout.visibility = 'visible'
            tanggal_akhir_widget.layout.visibility = 'visible'
            cek_mutasi_button.layout.visibility = 'visible'
        else:
            print("Username atau Password Anda salah.")

def cek_mutasi(b):
    with output_mutasi:
        clear_output()
        username = username_input.value
        password = password_input.value
        tanggal_awal = tanggal_awal_widget.value
        tanggal_akhir = tanggal_akhir_widget.value

        if not tanggal_awal or not tanggal_akhir:
            print("Harap pilih rentang tanggal yang valid.")
            return

        conn = sqlite3.connect('perbankan.db')
        c = conn.cursor()
        c.execute('''SELECT Transaksi.ID_Transaksi,
                            Transaksi.Tanggal_Transaksi,
                            Transaksi.Nominal_Transaksi,
                            Jenis_Transaksi.Jenis_Transaksi,
                            Biaya_Admin.Nominal_Biaya_Admin
                    FROM Transaksi
                    JOIN Biaya_Admin ON Transaksi.ID_Biaya_Admin = Biaya_Admin.ID_Biaya_Admin
                    JOIN Jenis_Transaksi ON Transaksi.ID_Jenis_Transaksi = Jenis_Transaksi.ID_Jenis_Transaksi
                    WHERE Transaksi.ID_Akun = (SELECT ID_Akun FROM Akun_MBanking WHERE Username = ?)
                    AND Transaksi.Tanggal_Transaksi BETWEEN ? AND ?''',
                  (username, tanggal_awal.strftime('%d-%m-%Y'), tanggal_akhir.strftime('%d-%m-%Y')))

        mutasi = c.fetchall()
        conn.close()

        if mutasi:
            df = pd.DataFrame(mutasi, columns=['ID Transaksi', 'Tanggal Transaksi',
                                               'Nominal Transaksi', 'Jenis Transaksi', 'Biaya Admin'])
            display(df)

            # Menghitung total debit dan kredit
            total_debit = df[df['Jenis Transaksi'] == 'Debit']['Nominal Transaksi'].sum()
            total_kredit = df[df['Jenis Transaksi'] != 'Debit']['Nominal Transaksi'].sum()
            selisih = total_debit - total_kredit

            # Menampilkan total debit dan kredit
            print(f"Total Debit: {total_debit}")
            print(f"Total Kredit: {total_kredit}")
            print(f"Selisih: {selisih}")
        else:
            print("Tidak ada transaksi dalam rentang tanggal tersebut.")

login_button.on_click(login)
cek_mutasi_button.on_click(cek_mutasi)

display(username_input, password_input, login_button, output_login, tanggal_awal_widget,
        tanggal_akhir_widget, cek_mutasi_button, output_mutasi)
```

## Cek Mutasi Berdasarkan Jenis Transaksi dan Tanggal

```python
# Fungsi untuk validasi user
def validasi_user(username, password):
    conn = sqlite3.connect('perbankan.db')
    c = conn.cursor()
    c.execute('SELECT * FROM Akun_MBanking WHERE Username = ? AND Password = ?', (username, password))
    result = c.fetchone()
    conn.close()
    return result is not None

# Fungsi untuk menangani login
def login(b):
    with output_login:
        clear_output()
        username = username_input.value
        password = password_input.value

        if validasi_user(username, password):
            print("Login berhasil. Silakan pilih jenis transaksi dan rentang tanggal.")
            jenis_transaksi_dropdown.layout.visibility = 'visible'
            tanggal_awal_picker.layout.visibility = 'visible'
            tanggal_akhir_picker.layout.visibility = 'visible'
            submit_button.layout.visibility = 'visible'
        else:
            print("Username atau Password Anda salah.")

# Widget input username dan password
username_input = widgets.Text(
    value='',
    description='Username:',
    style={'description_width': 'initial'})

password_input = widgets.Password(
    value='',
    description='Password:',
    style={'description_width': 'initial'})

login_button = widgets.Button(
    description='Login',
    button_style='success',
    tooltip='Klik untuk login')

output_login = widgets.Output()

login_button.on_click(login)

display(username_input, password_input, login_button, output_login)

# Widget untuk memilih jenis transaksi dengan label deskriptif
jenis_transaksi_dropdown = widgets.Dropdown(
    options=[
        ('Debit', 'T0'),
        ('Kredit Transfer Antar Rekening', 'T1'),
        ('Kredit Transfer Antar Bank', 'T2'),
        ('Kredit Transfer Virtual Account', 'T3'),
        ('Top-Up', 'T4')
    ],
    value='T0',
    description='Jenis Transaksi:',
    style={'description_width': 'initial'},
    layout={'visibility': 'hidden'})  # Awalnya widget ini tidak terlihat

# Widget untuk memilih tanggal awal dan akhir
tanggal_awal_picker = widgets.DatePicker(
    description='Tanggal Awal:',
    style={'description_width': 'initial'},
    layout={'visibility': 'hidden'})  # Awalnya widget ini tidak terlihat

tanggal_akhir_picker = widgets.DatePicker(
    description='Tanggal Akhir:',
    style={'description_width': 'initial'},
    layout={'visibility': 'hidden'})  # Awalnya widget ini tidak terlihat

# Tombol submit
submit_button = widgets.Button(
    description='Submit',
    button_style='info',
    tooltip='Klik untuk melihat mutasi transaksi',
    layout={'visibility': 'hidden'})  # Awalnya tombol ini tidak terlihat

output_mutasi = widgets.Output()

def submit(b):
    with output_mutasi:
        clear_output()
        jenis_transaksi = jenis_transaksi_dropdown.value
        tanggal_awal = tanggal_awal_picker.value
        tanggal_akhir = tanggal_akhir_picker.value

        if not tanggal_awal or not tanggal_akhir:
            print("Harap pilih rentang tanggal yang valid.")
            return

        conn = sqlite3.connect('perbankan.db')
        c = conn.cursor()
        query = '''SELECT Transaksi.ID_Transaksi,
                          Transaksi.Tanggal_Transaksi,
                          Transaksi.Nominal_Transaksi,
                          Jenis_Transaksi.Jenis_Transaksi,
                          Biaya_Admin.Nominal_Biaya_Admin
                     FROM Transaksi
                     JOIN Biaya_Admin ON Transaksi.ID_Biaya_Admin = Biaya_Admin.ID_Biaya_Admin
                     JOIN Jenis_Transaksi ON Transaksi.ID_Jenis_Transaksi = Jenis_Transaksi.ID_Jenis_Transaksi
                     WHERE Transaksi.ID_Akun = (SELECT ID_Akun FROM Akun_MBanking WHERE Username = ?)
                     AND Transaksi.Tanggal_Transaksi BETWEEN ? AND ?
                     AND Jenis_Transaksi.ID_Jenis_Transaksi = ?'''
        params = (username_input.value, tanggal_awal.strftime('%d-%m-%Y'),
                  tanggal_akhir.strftime('%d-%m-%Y'), jenis_transaksi)
        c.execute(query, params)

        mutasi = c.fetchall()
        conn.close()

        if mutasi:
            # Membuat DataFrame dari hasil query
            df = pd.DataFrame(mutasi, columns=['ID Transaksi', 'Tanggal Transaksi'
            , 'Nominal Transaksi', 'Jenis Transaksi', 'Biaya Admin'])
            display(df)
            total_debit = 0
            total_kredit = 0

            for index, row in df.iterrows():
                jenis_mutasi = 'Debit' if row['Jenis Transaksi'] in ['Debit'] else 'Kredit'
                if jenis_mutasi == 'Debit':
                    total_debit += row['Nominal Transaksi']
                else:
                    total_kredit += row['Nominal Transaksi']
            selisih = total_debit - total_kredit

            # Menampilkan total debit dan kredit
            print(f"Total Debit: {total_debit}")
            print(f"Total Kredit: {total_kredit}")
            print(f"Selisih: {selisih}")

        else:
            print("Tidak ada transaksi dalam rentang tanggal tersebut.")

login_button.on_click(login)
submit_button.on_click(submit)

display(jenis_transaksi_dropdown, tanggal_awal_picker, tanggal_akhir_picker, submit_button, output_mutasi)
```
